## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

## Data Wrangling

### Gathering Data

#### Load df ...

```python
df = pd.read_csv("PRSA_Data_Changping_20130301-20170228.csv")
df.head()
```

**Insight:**
- Dataset berhasil dimuat, terdiri 18 kolom.
- Data ini memuat catatan historis kualitas udara dan cuaca dari stasiun pemantauan Changping.

### Assessing Data

#### Identifying Missing Values dan Tipe Data

```python
# Mengecek tipe data dan missing value
print(df.info())
print("\nJumlah Missing Values:")
print(df.isna().sum())

# Mengecek data duplikat
print("\nJumlah Data Duplikat: ", df.duplicated().sum())
```

**Steps to Take:**
- Terdapat *missing value* (nilai kosong) pada beberapa kolom polutan (seperti PM2.5, PM10, SO2, dll) dan cuaca. Karena ini data *time-series*, kita akan menanganinya menggunakan metode interpolasi.
- Terdapat nilai kosong pada kolom arah angin (`wd`). Akan diisi dengan nilai yang paling sering muncul (modus).
- Kolom waktu (`year`, `month`, `day`, `hour`) masih terpisah dalam format *integer*. Akan digabungkan menjadi satu kolom bertipe `datetime`.

**Insight:**
- Ditemukan banyak *missing value* (data kosong) pada kolom kualitas udara, contohnya kolom PM2.5 memiliki 774 baris kosong.
- Kolom waktu (`year`, `month`, `day`, `hour`) masih terpisah dalam format integer sehingga kurang ideal untuk analisis *time-series*.
- Tidak ada indikasi data duplikat (*0 duplicated rows*).

### Cleaning Data

#### Fixing Missing Values & Formatting Datetime

```python
# Menggabungkan kolom tahun, bulan, hari, dan jam menjadi tipe datetime
df['datetime'] = pd.to_datetime(df[['year', 'month', 'day', 'hour']])

# Menangani Missing Value menggunakan interpolasi linier untuk kolom numerik
kolom_numerik = df.select_dtypes(include=['float64', 'int64']).columns
df[kolom_numerik] = df[kolom_numerik].interpolate(method='linear', limit_direction='both')

# Mengisi missing value pada kolom kategorik 'wd' (arah angin) dengan modus
df['wd'] = df['wd'].fillna(df['wd'].mode()[0])

# Cek kembali apakah masih ada missing value
print("Sisa Missing Value:\n", df.isna().sum())
```

**Insight:**
- Proses *cleaning* berhasil; seluruh *missing value* numerik sudah diisi secara mulus menggunakan metode interpolasi linier.
- Kolom kategori yang kosong (`wd` atau arah angin) telah terisi otomatis dengan nilai modus.
- Kolom waktu sudah digabungkan menjadi satu kesatuan berformat `datetime`. Total missing value sekarang adalah 0.

## Exploratory Data Analysis (EDA)

### Explore Tren Bulanan dan Pola Harian PM2.5

```python
# Eksplorasi Pertanyaan 1: Rata-rata bulanan tiap tahun
df['year_month'] = df['datetime'].dt.to_period('M')
monthly_pm25 = df.groupby('year_month')['PM2.5'].mean().reset_index()
monthly_pm25['year_month'] = monthly_pm25['year_month'].astype(str)

# Eksplorasi Pertanyaan 2: Rata-rata per jam dalam sehari
hourly_pm25 = df.groupby('hour')['PM2.5'].mean().reset_index()

# Tampilkan beberapa baris eksplorasi tren harian
display(hourly_pm25.sort_values(by='PM2.5', ascending=False).head())
```

**Insight:**
- Berdasarkan hasil agregasi *grouping*, nilai rata-rata PM2.5 bulanan menunjukkan fluktuasi yang berulang setiap pergantian tahun.
- Pada tabel agregasi per jam, angka PM2.5 terlihat menurun di siang bolong dan secara konsisten merangkak naik saat sore beralih ke malam hari.

## Visualization & Explanatory Analysis

### Pertanyaan 1:

```python
plt.figure(figsize=(15, 6))
sns.lineplot(data=monthly_pm25, x='year_month', y='PM2.5', marker='o', color='#E63946')
plt.title('Tren Rata-rata Konsentrasi PM2.5 Bulanan di Changping (2013-2017)', fontsize=16)
plt.xlabel('Tahun-Bulan', fontsize=12)
plt.ylabel('Rata-rata PM2.5 (µg/m³)', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

### Pertanyaan 2:

```python
plt.figure(figsize=(10, 5))
sns.barplot(data=hourly_pm25, x='hour', y='PM2.5', palette='viridis', hue='hour', legend=False)
plt.title('Rata-rata Konsentrasi PM2.5 Berdasarkan Jam di Changping', fontsize=16)
plt.xlabel('Jam dalam Sehari (0-23)', fontsize=12)
plt.ylabel('Rata-rata PM2.5 (µg/m³)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

**Insight:**
- **Grafik Tren Bulanan:** Pola musiman tergambar sangat jelas. Lonjakan garis merah (PM2.5 tertinggi) selalu bertepatan dengan bulan-bulan akhir tahun hingga awal tahun berikutnya.
- **Grafik Pola Harian:** Tingkat polusi membentuk kurva "U". Polusi sangat pekat di malam-dini hari, lalu perlahan turun dan paling bersih di kisaran pukul 14.00 - 16.00 sore.
